# 00 Merge Source Tables

Merges the two working spreadsheets into the per-subcorpus source tables.

**Inputs:** `kid_lit_all_ru.xlsx`, `kid_lit_all_foreign.xlsx`  
**Outputs:** `kid_lit_100_ru.csv`, `kid_lit_100_foreign.csv`

> **Not re-runnable from this repository.** This notebook reads the running text of the 100 books, which is under copyright and is not distributed. It is included as a record of how the released metrics were produced.


```python
# --- Paths ---------------------------------------------------------------
# BASE_PATH must point at a directory holding this notebook's input files.
# Set the KIDLIT_BASE environment variable, or edit the fallback below.
# In Google Colab: mount Drive first, then set KIDLIT_BASE to the folder there.
import os
BASE_PATH = os.environ.get("KIDLIT_BASE", "../data/") + "/"
```

```python
import pandas as pd
```

```python
df_ru = pd.read_excel(BASE_PATH + '/Обработка данных/kid_lit_all_ru.xlsx')
df_fo = pd.read_excel(BASE_PATH + '/Обработка данных/kid_lit_all_foreign.xlsx')
```

```python
with open(BASE_PATH + "/Обработка данных/Клубника бегает трусцой.txt", "r", encoding="utf-8") as file:
    content = file.read()

df_ru.loc[df_ru['title'] == 'Клубника бегает трусцой', 'text'] = content
```

```python
df_ru.to_csv(BASE_PATH + '/kid_lit_100_ru.csv', index=False, sep=';', encoding='utf-8-sig')
df_fo.to_csv(BASE_PATH + '/kid_lit_100_foreign.csv', index=False, sep=';', encoding='utf-8-sig')
```

```python
print(f"✅ Русских книг: {len(df_ru)}, с текстом: {df_ru['text'].notna().sum()}")
print(f"✅ Иностранных книг: {len(df_fo)}, с текстом: {df_fo['text'].notna().sum()}")
```
